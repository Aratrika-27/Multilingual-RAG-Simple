# Multilingual Code-Mix Chatbot Demo

This notebook demonstrates the multilingual code-mix chatbot with RAG capabilities.

```python
import sys
import os
from pathlib import Path

# Add the project root to the Python path
project_root = Path().absolute().parent
sys.path.insert(0, str(project_root))
```

## Load Sample Data

```python
from src.loader.data_loader import load_sample_data

# Load sample data
documents = load_sample_data()

# Display the documents
for i, doc in enumerate(documents):
    print(f"Document {i+1}:")
    print(f"Content: {doc.page_content}")
    print(f"Metadata: {doc.metadata}")
    print()
```

## Initialize RAG Pipeline

```python
from src.rag_pipeline import initialize_rag

# Initialize RAG pipeline
rag = initialize_rag(documents=documents)
```

## Test Queries

```python
# Test queries
test_queries = [
    "What's the weather like tomorrow?",
    "कल का मौसम कैसा होगा?",  # Hindi: What will the weather be like tomorrow?
    "আগামীকাল আবহাওয়া কেমন হবে?",  # Bengali: What will the weather be like tomorrow?
    "Kal ka weather kaisa hoga?",  # Hindi-English mix
    "Doctor appointment के लिए reminder set करो",  # Hindi-English mix
    "Kolkata में best restaurants কোনগুলি?"  # Hindi-Bengali-English mix
]

for query in test_queries:
    print(f"\nQuery: {query}")
    try:
        result = rag.query(query)
        print(f"Response: {result['response']}")
        print(f"Retrieved context: {result['context'][:200]}...")
    except Exception as e:
        print(f"Error: {str(e)}")
```

## Interactive Demo

```python
from IPython.display import display, HTML

def chat_interface():
    query = input("You: ")
    if query.lower() in ["exit", "quit"]:
        return
    
    try:
        result = rag.query(query)
        print(f"Chatbot: {result['response']}")
    except Exception as e:
        print(f"Error: {str(e)}")
    
    # Continue the conversation
    chat_interface()

print("===== Multilingual Code-Mix Chatbot =====")
print("Type 'exit' or 'quit' to end the conversation.")
chat_interface()
```
